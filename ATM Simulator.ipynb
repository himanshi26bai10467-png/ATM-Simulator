```python
# ATM SIMULATOR

balance = 10000
transactions = []


# LANGUAGE SELECTION
def select_language():
    print("\n========== LANGUAGE ==========")
    print("1. English")
    print("2. Hindi")

    choice = input("Select language / Bhasha chune: ")

    if choice == "1":
        return "English"
    elif choice == "2":
        return "Hindi"
    else:
        print("Invalid choice! English selected by default.")
        return "English"

# ==========================================
# ATM SIMULATOR MINI PROJECT (WITH HINDI)
# ==========================================

# 1. Sabhi bhashao ke messages ki Dictionary
MESSAGES = {
    "English": {
        "lang_default": "Invalid choice! English selected by default.",
        "welcome": "\nWELCOME TO ATM SIMULATOR",
        "select_account": "========== ACCOUNT TYPE ==========",
        "savings": "1. Savings Account",
        "current": "2. Current Account",
        "enter_account_choice": "Select account type: ",
        "invalid_account": "Invalid choice! Savings Account selected by default.",
        "menu_header": "========== MAIN MENU ==========",
        "menu_balance": "1. Check Balance",
        "menu_deposit": "2. Deposit Money",
        "menu_withdraw": "3. Withdraw Money",
        "menu_exit": "4. Exit",
        "enter_menu_choice": "Choose an option: ",
        "invalid_menu": "Invalid option! Please try again.",
        "current_balance": "Your current balance is: ₹",
        "enter_deposit": "Enter amount to deposit: ₹",
        "deposit_success": "Amount deposited successfully!",
        "enter_withdraw": "Enter amount to withdraw: ₹",
        "withdraw_success": "Please collect your cash!",
        "insufficient_balance": "Insufficient balance!",
        "invalid_amount": "Please enter a valid positive amount.",
        "thank_you": "Thank you for using our ATM! Have a nice day."
    },
    "Hindi": {
        "lang_default": "Anya vikalp chuna gaya! English swatah chuni gayi.",
        "welcome": "\nATM SIMULATOR ME AAPKA SWAGAT HAI",
        "select_account": "========== KHATA KA PRAKAR ==========",
        "savings": "1. Bachat Khata (Savings)",
        "current": "2. Chalu Khata (Current)",
        "enter_account_choice": "Khata ka prakar chune: ",
        "invalid_account": "Galat vikalp! Swatah Bachat Khata chuna gaya.",
        "menu_header": "========== MUKHYA MENU ==========",
        "menu_balance": "1. Balance Dekhein (Check Balance)",
        "menu_deposit": "2. Paise Jama Karein (Deposit)",
        "menu_withdraw": "3. Paise Nikalein (Withdraw)",
        "menu_exit": "4. Bahar Nikalein (Exit)",
        "enter_menu_choice": "Koyi ek vikalp chune: ",
        "invalid_menu": "Galat vikalp! Kripya dobara koshish karein.",
        "current_balance": "Aapka kul balance hai: ₹",
        "enter_deposit": "Jama karne ke liye rashi dalein: ₹",
        "deposit_success": "Rashi saphaltapoorvak jama ho gayi!",
        "enter_withdraw": "Nikalne ke liye rashi dalein: ₹",
        "withdraw_success": "Kripya apna cash le lein!",
        "insufficient_balance": "Khate me paryapt balance nahi hai!",
        "invalid_amount": "Kripya sahi rashi dalein.",
        "thank_you": "ATM ka upyog karne ke liye dhanyawad! Aapka din shubh ho."
    }
}



# 3. Account Type Selection Function
def select_account_type(lang):
    msg = MESSAGES[lang]
    
    print("\n" + msg["select_account"])
    print(msg["savings"])
    print(msg["current"])
    
    choice = input(msg["enter_account_choice"])
    
    if choice == "1":
        return "Savings Account"
    elif choice == "2":
        return "Current Account"
    else:
        print(msg["invalid_account"])
        return "Savings Account"


# 4. Main ATM Logic
def main_atm():
    # Balance initialize karein
    balance = 5000.0  # Initial Balance
    
    # Language select karein
    selected_lang = select_language()
    msg = MESSAGES[selected_lang]
    
    # Account Type select karein
    account_type = select_account_type(selected_lang)
    
    print(msg["welcome"])
    print(f"Selected Account: {account_type}")

    while True:
        print("\n" + msg["menu_header"])
        print(msg["menu_balance"])
        print(msg["menu_deposit"])
        print(msg["menu_withdraw"])
        print(msg["menu_exit"])
        
        choice = input(msg["enter_menu_choice"])
        
        # Balance Check
        if choice == "1":
            print(f"\n{msg['current_balance']}{balance:.2f}")
            
        # Deposit Money
        elif choice == "2":
            try:
                amount = float(input(msg["enter_deposit"]))
                if amount > 0:
                    balance += amount
                    print(msg["deposit_success"])
                    print(f"{msg['current_balance']}{balance:.2f}")
                else:
                    print(msg["invalid_amount"])
            except ValueError:
                print(msg["invalid_amount"])
                
        # Withdraw Money
        elif choice == "3":
            try:
                amount = float(input(msg["enter_withdraw"]))
                if amount <= 0:
                    print(msg["invalid_amount"])
                elif amount > balance:
                    print(msg["insufficient_balance"])
                else:
                    balance -= amount
                    print(msg["withdraw_success"])
                    print(f"{msg['current_balance']}{balance:.2f}")
            except ValueError:
                print(msg["invalid_amount"])
                
        # Exit
        elif choice == "4":
            print(f"\n{msg['thank_you']}")
            break
            
        else:
            print(msg["invalid_menu"])

# Main program run karne ke liye
if __name__ == "__main__":
    main_atm()

# ACCOUNT TYPE
def select_account_type():
    print("\n========== ACCOUNT TYPE ==========")
    print("1. Savings Account")
    print("2. Current Account")

    choice = input("Select account type: ")

    if choice == "1":
        return "Savings Account"
    elif choice == "2":
        return "Current Account"
    else:
        print("Invalid choice! Savings Account selected by default.")
        return "Savings Account"


# CHECK BALANCE
def check_balance():
    print("\nYour balance is:", balance)


# DEPOSIT
def deposit():
    global balance

    amount = int(input("Enter amount to deposit: "))

    if amount > 0:
        balance = balance + amount
        transactions.append("Deposited: " + str(amount))

        print("Amount deposited successfully.")
        print("New balance:", balance)

    else:
        print("Invalid amount.")


# WITHDRAW
def withdraw():
    global balance

    amount = int(input("Enter amount to withdraw: "))

    if amount > balance:
        print("Insufficient balance.")

    elif amount <= 0:
        print("Invalid amount.")

    else:
        balance = balance - amount
        transactions.append("Withdrawn: " + str(amount))

        print("Please collect your cash.")
        print("Remaining balance:", balance)


# TRANSFER
def transfer():
    global balance

    receiver = input("Enter receiver account number: ")
    amount = int(input("Enter amount to transfer: "))

    if len(receiver) != 10:
        print("Invalid account number.")

    elif amount <= 0:
        print("Invalid amount.")

    elif amount > balance:
        print("Insufficient balance.")

    else:
        balance = balance - amount
        transactions.append("Transferred: " + str(amount))

        print("Money transferred successfully.")
        print("Remaining balance:", balance)


# MAIN PROGRAM

print("================================")
print("       WELCOME TO ATM")
print("================================")

language = select_language()

print("\nSelected Language:", language)

account_type = select_account_type()

print("Selected Account:", account_type)


# ATM MENU
while True:

    print("\n========== ATM MENU ==========")
    print("1. Check Balance")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Transfer Money")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        check_balance()

    elif choice == "2":
        deposit()

    elif choice == "3":
        withdraw()

    elif choice == "4":
        transfer()

    elif choice == "5":
        print("\nThank you for using ATM.")
        break

    else:
        print("Invalid choice.")
```
